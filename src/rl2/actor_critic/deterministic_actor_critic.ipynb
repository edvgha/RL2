**Deterministic policy gradient (deterministic actor-critic)**

**Initialization:** A given behavior policy $\beta(a \mid s)$. A deterministic target policy $\mu(s, \theta_0)$, where $\theta_0$ is the initial parameter. A value function $q(s, a, w_0)$, where $w_0$ is the initial parameter. $\alpha_w, \alpha_\theta > 0$.

**Goal:** Learn an optimal policy to maximize $J(\theta)$.

At time step $t$ in each episode, do:

1. Generate $a_t$ following $\beta$, then observe $r_{t+1}, s_{t+1}$.

2. **TD error:**
$$\delta_t = r_{t+1} + \gamma\, q\bigl(s_{t+1}, \mu(s_{t+1}, \theta_t), w_t\bigr) - q(s_t, a_t, w_t)$$

3. **Actor (policy update):**
$$\theta_{t+1} = \theta_t + \alpha_\theta\, \nabla_\theta \mu(s_t, \theta_t)\, \bigl.\nabla_a q(s_t, a, w_t)\bigr|_{a = \mu(s_t)}$$

4. **Critic (value update):**
$$w_{t+1} = w_t + \alpha_w\, \delta_t\, \nabla_w q(s_t, a_t, w_t)$$
