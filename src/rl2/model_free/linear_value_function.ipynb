## TD Learning of State Values with Function Approximation

### **Initialization**
* A function $\hat{v}(s,w)$ that is differentiable in $w$.
* Initial parameter $w_0$.

**Goal:** Learn the true state values of a given policy $\pi$.

---

### **Algorithm Body**

For each episode $\{(s_t, r_{t+1}, s_{t+1})\}_t$ generated by $\pi$, **do**:
* For each sample $(s_t, r_{t+1}, s_{t+1})$, **do**:
    
    * **In the general case:**
        $$w_{t+1} = w_t + \alpha_t \left[ r_{t+1} + \gamma \hat{v}(s_{t+1}, w_t) - \hat{v}(s_t, w_t) \right] \nabla_w \hat{v}(s_t, w_t)$$
        
    * **In the linear case:**
        $$w_{t+1} = w_t + \alpha_t \left[ r_{t+1} + \gamma \phi^T(s_{t+1})w_t - \phi^T(s_t)w_t \right] \phi(s_t)$$
