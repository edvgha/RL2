### MC $\epsilon$-Greedy (A Variant of MC Exploring Starts)

* **Initialization:** * Initial policy $\pi_0(a \mid s)$ and initial value $q(s,a)$ for all $(s,a)$.
  * $\text{Returns}(s,a) = 0$ and $\text{Num}(s,a) = 0$ for all $(s,a)$.
  * Exploration parameter $\epsilon \in (0,1]$.
* **Goal:** Search for an optimal policy.

---

**For each episode, do:**

#### 1. Episode Generation
* Select an arbitrary starting state-action pair $(s_0, a_0)$ (the *exploring starts* condition is not strictly required). 
* Following the current policy $\pi$, generate an episode of length $T$: 
  $$s_0, a_0, r_1, s_1, a_1, r_2, \dots, s_{T-1}, a_{T-1}, r_T$$

#### 2. Episode Processing
* **Initialize return accumulator:** $g \leftarrow 0$
* **For each step of the episode**, $t = T-1, T-2, \dots, 0$, **do**:
  * $g \leftarrow \gamma g + r_{t+1}$
  * $\text{Returns}(s_t, a_t) \leftarrow \text{Returns}(s_t, a_t) + g$
  * $\text{Num}(s_t, a_t) \leftarrow \text{Num}(s_t, a_t) + 1$
  
  ##### Policy Evaluation:
  * $q(s_t, a_t) \leftarrow \frac{\text{Returns}(s_t, a_t)}{\text{Num}(s_t, a_t)}$
  
  ##### Policy Improvement:
  * Let $a^* = \arg\max_{a} q(s_t, a)$
  * Update the policy profile for all $a \in A(s_t)$:
    $$\pi(a \mid s_t) = \begin{cases} 1 - \frac{|A(s_t)| - 1}{|A(s_t)|}\epsilon & \text{if } a = a^* \\ \frac{1}{|A(s_t)|}\epsilon & \text{if } a \neq a^* \end{cases}$$
