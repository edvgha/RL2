### Policy Iteration Algorithm

* **Initialization:** The system model, $p(r \mid s,a)$ and $p(s' \mid s,a)$ for all $(s,a)$, is known. An initial policy guess $\pi_0$ is provided.
* **Goal:** Search for the optimal state value and an optimal policy.

---

**While** $v^{\pi_k}$ has not converged, for the $k$-th iteration, **do**:

#### 1. Policy Evaluation
* **Initialization:** Choose an arbitrary initial value function guess $v_{\pi_k}^{(0)}$.
* **While** $v_{\pi_k}^{(j)}$ has not converged, for the $j$-th iteration, **do**:
  * **For every state** $s \in S$, **do**:
    $$v_{\pi_k}^{(j+1)}(s) = \sum_{a} \pi_k(a \mid s) \left[ \sum_{r} p(r \mid s,a)r + \gamma \sum_{s'} p(s' \mid s,a)v_{\pi_k}^{(j)}(s') \right]$$

#### 2. Policy Improvement
* **For every state** $s \in S$, **do**:
  * **For every action** $a \in A$, **do**:
    $$q^{\pi_k}(s,a) = \sum_{r} p(r \mid s,a)r + \gamma \sum_{s'} p(s' \mid s,a)v^{\pi_k}(s')$$
  * **Determine the greedy action:** $$a^*_k(s) = \arg\max_{a} q^{\pi_k}(s,a)$$
  * **Update the policy:** $$\pi_{k+1}(a \mid s) = \begin{cases} 1 & \text{if } a = a^*_k(s) \\ 0 & \text{otherwise} \end{cases}$$
