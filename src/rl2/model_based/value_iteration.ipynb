### Value Iteration Algorithm

* **Initialization:** The probability models $p(r \mid s, a)$ and $p(s' \mid s, a)$ for all $(s, a)$ are known.
* **Initial Guess:** $v_0$
* **Goal:** Search for the optimal state value and an optimal policy for solving the Bellman optimality equation.

---

**While** $v_k$ has not converged (i.e., $\|v_k - v_{k-1}\|$ is greater than a predefined small threshold $\theta$), for the $k$-th iteration, **do**:

1. **For every state** $s \in S$, **do**:
   * **For every action** $a \in A(s)$, **do**:
     * **$q$-value calculation:** $$q_k(s, a) = \sum_{r} p(r \mid s, a) \cdot r + \gamma \sum_{s'} p(s' \mid s, a) \cdot v_k(s')$$
     * **Maximum action selection:** $$a^*_k(s) = \arg\max_{a} q_k(s, a)$$
     * **Policy update:** $$\pi_{k+1}(a \mid s) = \begin{cases} 1 & \text{if } a = a^*_k(s) \\ 0 & \text{otherwise} \end{cases}$$
     * **Value update:** $$v_{k+1}(s) = \max_{a} q_k(s, a)$$
