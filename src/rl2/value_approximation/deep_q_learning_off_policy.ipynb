## Deep Q-learning (Off-policy Version)

### **Initialization**
* A main network with parameter $w$ and a target network with parameter $w^T$, initialized with the same values ($w^T = w$).

**Goal:** Learn an optimal target network to approximate the optimal action values from the experience samples generated by a given behavior policy $\pi_b$.

---

### **Algorithm Body**

Store the experience samples generated by $\pi_b$ in a replay buffer $\mathcal{B} = \{(s, a, r, s')\}$.

For each iteration, **do**:
1. Uniformly draw a mini-batch of samples from $\mathcal{B}$.
2. For each sample $(s, a, r, s')$, calculate the target value:
   $$y^T = r + \gamma \max_{a \in A(s')} \hat{q}(s', a, w^T)$$
   *(where $w^T$ is the parameter of the target network)*
3. Update the main network parameter $w$ to minimize the loss:
   $$L(w) = \left( y^T - \hat{q}(s, a, w) \right)^2$$
   using the mini-batch of samples.
4. Set $w^T = w$ every $C$ iterations.
