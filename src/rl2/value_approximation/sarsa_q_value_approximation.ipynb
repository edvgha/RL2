## Sarsa with Function Approximation

### **Initialization**
* Initial parameter $w_0$.
* Initial policy $\pi_0$.
* $\alpha_t = \alpha > 0$ for all $t$.
* $\epsilon \in (0,1)$.

**Goal:** Learn an optimal policy that can lead the agent to the target state from an initial state $s_0$.

---

### **Algorithm Body**

For each episode, **do**:
1. Generate $a_0$ at $s_0$ following $\pi_0(s_0)$
2. If $s_t$ ($t = 0, 1, 2, \dots$) is not the target state, **do**:
    * **Collect** the experience sample $(r_{t+1}, s_{t+1}, a_{t+1})$ given $(s_t, a_t)$: 
        * Generate $r_{t+1}, s_{t+1}$ by interacting with the environment.
        * Generate $a_{t+1}$ following $\pi_t(s_{t+1})$.
    * **Update q-value**:
        $$w_{t+1} = w_t + \alpha_t \left[ r_{t+1} + \gamma \hat{q}(s_{t+1}, a_{t+1}, w_t) - \hat{q}(s_t, a_t, w_t) \right] \nabla_w \hat{q}(s_t, a_t, w_t)$$
    * **Update policy**:
        $$\pi_{t+1}(a|s_t) = 1 - \frac{\epsilon(|A(s_t)|-1)}{|A(s_t)|} \quad \text{if } a = \arg\max_{a \in A(s_t)} \hat{q}(s_t, a, w_{t+1})$$
        $$\pi_{t+1}(a|s_t) = \frac{\epsilon}{|A(s_t)|} \quad \text{otherwise}$$
    * **Transition**:
        $$s_t \leftarrow s_{t+1}, \quad a_t \leftarrow a_{t+1}$$
